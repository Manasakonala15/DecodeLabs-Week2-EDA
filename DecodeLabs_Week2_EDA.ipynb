```
import pandas as pd

# Load Dataset
df = pd.read_excel("Dataset for Data Analytics(week2).xlsx")

print("="*50)
print("DATASET OVERVIEW")
print("="*50)

print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))
print("\nColumns:")
print(df.columns.tolist())

# --------------------------
# BASIC STATISTICS
# --------------------------

print("\n")
print("="*50)
print("BASIC STATISTICS")
print("="*50)

print(df.describe())

# --------------------------
# BUSINESS METRICS
# --------------------------

print("\n")
print("="*50)
print("BUSINESS METRICS")
print("="*50)

print("Total Revenue:", round(df['TotalPrice'].sum(),2))
print("Average Order Value:", round(df['TotalPrice'].mean(),2))
print("Median Order Value:", round(df['TotalPrice'].median(),2))
print("Unique Customers:", df['CustomerID'].nunique())

# --------------------------
# PRODUCT ANALYSIS
# --------------------------

print("\n")
print("="*50)
print("TOP PRODUCTS")
print("="*50)

print(df['Product'].value_counts())

print("\nRevenue by Product:")
print(df.groupby('Product')['TotalPrice'].sum().sort_values(ascending=False))

# --------------------------
# PAYMENT METHOD ANALYSIS
# --------------------------

print("\n")
print("="*50)
print("PAYMENT METHODS")
print("="*50)

print(df['PaymentMethod'].value_counts())

# --------------------------
# ORDER STATUS ANALYSIS
# --------------------------

print("\n")
print("="*50)
print("ORDER STATUS")
print("="*50)

print(df['OrderStatus'].value_counts())

# --------------------------
# REFERRAL SOURCE ANALYSIS
# --------------------------

print("\n")
print("="*50)
print("REFERRAL SOURCES")
print("="*50)

print(df['ReferralSource'].value_counts())

# --------------------------
# OUTLIER CHECK
# --------------------------

print("\n")
print("="*50)
print("OUTLIER ANALYSIS")
print("="*50)

Q1 = df['TotalPrice'].quantile(0.25)
Q3 = df['TotalPrice'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = df[(df['TotalPrice'] < lower) | (df['TotalPrice'] > upper)]

print("Number of Outliers:", len(outliers))

# --------------------------
# OBSERVATIONS
# --------------------------

print("\nEDA COMPLETED SUCCESSFULLY")
```
